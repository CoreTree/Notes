## Easy
### Valid Anagram
Given two strings `s` and `t`, return `true` if the two strings are anagrams of each other, otherwise return `false`.

An **anagram** is a string that contains the exact same characters as another string, but the order of the characters can be different.

#### Example 1
```
Input: s = "racecar", t = "carrace"

Output: true
```
#### Example 2
```
Input: s = "jar", t = "jam"

Output: false
```
#### Contraints
- `s` and `t` consist of lowercase English letters

````python
from collections import defaultdict

class Solution:
    def is_anagram(self, s:str, t:str) -> bool:
        lookup = defaultdict(int)
        for ltr in s:
            lookup[ltr] = lookup.get(ltr, 0) + 1
        for ltr in t:
            if ltr not in lookup:
                return False
            lookup[ltr] -= 1
            if lookup[ltr] == 0:
                del lookup[ltr]
        return len(lookup) == 0

if __name__ == "__main__":
    test_case_1_s = "racecar"
    test_case_1_t = "carrace"
    test_case_1_expected = True

    test_case_2_s = "jar"
    test_case_2_t = "jam"
    test_case_2_expected = False

    test_case_3_s = "sandrams"
    test_case_3_t = "mansards"
    test_case_3_expected = True

    sol = Solution()

    assert sol.is_anagram(test_case_1_s, test_case_1_t) == test_case_1_expected
    print("Test 1 passed.")
    assert sol.is_anagram(test_case_2_s, test_case_2_t) == test_case_2_expected
    print("Test 2 passed.")
    assert sol.is_anagram(test_case_3_s, test_case_3_t) == test_case_3_expected
    print("All Tests Passed.")
````

## Medium
### Group Anagrams
Given an array of strings `strs`, group all anagrams together into sublists. You may return the output in **any order**.abs
- An anagram is a string that contains the exact same characters as another string, but the order of the characters can be different.

### Example 1;
- input: strs = ["act", "pots", "tops", "cat", "stop", "hat"]
- output: [["hat"], ["act", "cat"], ["stop", "pots", "tops"]]

### Constraints
- `1 <= strs.length <= 1000`
- `0 <= strs[i].length <= 100`
- `strs[i]` is made up of lowercase English letters

````python
from collections import defaultdict

class Solution:
    def groups(self, strs:list[str]) -> list[list[str]]:
        res = defaultdict(list)
        for s in strs:
            count = [0] * 26
            for c in s:
                count[ord(c) - ord('a')] += 1
            res[tuple(count)].append(s)
        return res.values()

if __name__ == "__main__":
    sol = Solution()
    
    #Test 1
    test_list_1 = ["act", "pots", "tops", "cat", "stop", "hat"]
    test_res_1 = [["hat"], ["act", "cat"], ["pots", "stop", "tops"]]

    temp = list(sol.groups(test_list_1))
    assert len(temp) == len(test_res_1)
    for i in range(len(temp)):
        temp[i].sort()
        assert test_res_1.index(temp[i]) > -1
    print("Test 1 passed.")
````

#### Example 1

```
Input: height = [0,1,0,2,1,0,1,3,2,1,2,1]
Output: 6
Explanation: The above elevation map (black section) is represented by array `height`. In this case, 6 units of rain water (blue section) are being trapped
```

#### Example 2
```
Input: height = [4,2,0,3,2,5]
Output: 9
```

#### Constraints
- `n == height.length`
- `1 <= n <= 2 * (10 ^ 4)`
- `0 <= height[i] <= (10^5)`

````python
class Solution:
    def trapping_rainwater_n(self, heights:list[int]) -> int: #O(n) runtime
        res = 0
        max_left = [0]
        max_right = [0]
        for i in range(1,len(heights)):
            max_left.append(max(heights[i-1], max_left[-1]))
        for i in range(len(heights) - 1, 0, -1):
            max_right.append(max(heights[i-1], max_right[-1]))
        max_right = max_right[::-1] #Reverse list order because we appended left->right and this should be right->left
        print("Max_left:", max_left)
        print("Max_right:", max_right)
        diff = [max_left[i] - max_right[i] for i in range(len(max_left))]
        print("Diff:", diff)
        for i,val in enumerate(heights):
            print(i, val, diff[i] - val)
            res += max(0, diff[i] - val)
        return res

if __name__ == "__main__":
    test_case_1_height = [0,1,0,2,1,0,1,3,2,1,2,1]
    test_case_1_expected = 6

    test_case_2_height = [4,2,0,3,2,5]
    test_case_2_output = 9

    sol = Solution()
    test_case_1_res = sol.trapping_rainwater_n(test_case_1_height)
    print("Test Case #1:", test_case_1_res, test_case_1_expected)
    assert test_case_1_res == test_case_1_expected
    assert sol.trapping_rainwater_n(test_case_2_height) == test_case_2_expected
    assert sol.trapping_rainwater_n(test_case_3_height) == test_case_3_expected
````
