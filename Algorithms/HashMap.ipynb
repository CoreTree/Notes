```python
# Valid Anagrams
def is_anagram(s:str, t:str) -> bool:
    s_dict = {}
    for ch in s:
        s_dict[ch] = s_dict.get(ch, 0) + 1
    t_dict = {}
    for ch in t:
        t_dict[ch] = t_dict.get(ch, 0) + 1
    for ch in t_dict:
        if ch not in s_dict:
            return False
        if s_dict[ch] != t_dict[ch]:
            return False
        del s_dict[ch]
    return len(s_dict) == 0
```

## Contains Duplicate II
Given an integer array nums and an integer k, return true if there are two distinct indices (`i and j`) in the array such that `nums[i] == nums[j]` and `abs(i - j) <= k`.
### Example 1
Input:
- `nums = [1,2,3,1]`
- `k = 3`
Output: true

### Example 2
Input:
- `nums = [1,0,1,1]`
- `k = 1`
Output: true

### Example 3
Input:
- `nums = [1, 2, 3, 1, 2, 3]`
- `k = 2`
Output: false

```python
class Solution:
    def containsNearbyDuplicate(self, nums:list[int], k:int) -> bool:
        lookup = {}
        for i,v in enumerate(nums):
            if str(v) not in lookup:
                lookup.update({str(v): i})
            else:
                if abs(i - lookup[str(v)]) <= k:
                    return True
                lookup[str(v)] = i
        return False
```

```python
def test_testcase(numb:int, nums:list[int], k:int, expected:bool):
    sol = Solution()
    res = sol.containsNearbyDuplicate(nums, k)
    print(f"Test Case {numb}:")
    print("\tResults:", res)
    assert res == expected
    print("\tExpected:", expected)
```

```python
test_case_1_nums = [1, 2, 3, 1]
test_case_1_k = 3
test_case_1_expected = True
test_testcase(1,
              test_case_1_nums,
              test_case_1_k,
              test_case_1_expected)

test_case_2_nums = [1, 0, 1, 1]
test_case_2_k = 1
test_case_2_expected = True
test_testcase(2,
              test_case_2_nums,
              test_case_2_k,
              test_case_2_expected)

test_case_3_nums = [1, 2, 3, 1, 2, 3]
test_case_3_k = 2
test_case_3_expected = False
test_testcase(3,
              test_case_3_nums,
              test_case_3_k,
              test_case_3_expected)
```
