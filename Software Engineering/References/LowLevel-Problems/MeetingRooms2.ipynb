# Meeting Rooms II
## Problem
Given an array of meeting time itnerval objects consisting of start and end times `[[start_1, end_1], [start_2, end_2], ...] (start_i, end_i)`, find minimum number of rooms required to schedule all meetings without any conflicts.

Note: `(0,8),(8,10)` is **NOT** considered a conflict at 8.

## Example 1
```
Input: intervals = [(0,40), (5,10), (15,20)]
Output: 2
```
Explanation:
- day1: `(0, 40)`
- day2: `(5, 10), (15, 20)`
## Example 2
```
Input: intervals = [(4,9)]
Output: 1
```
## Constraints

- ``

````python
class Interval:
    def __init__(self, start:int, end:int):
        self.start = start
        self.end = end

class Solution:
    def minMeetingRooms(self, intervals: List[Interval]) -> int:
        start = [i.start for i in intervals]
        end = [i.end for i in intervals]
        start.sort()
        end.sort()

        maximum = 0
        concurrent = 0
        start_i = 0
        end_i = 0

        while start_i < len(start) and end_i < len(end):
            if start[start_i] < end[end_i]:
                concurrent += 1
                start_i += 1
            elif start[start_i] > end[end_i]:
                concurrent -= 1
                end_i += 1
            else:
                start_i += 1
                end_i += 1
            maximum = max(maximum, concurrent)
        return maximum

if __name__ == "__main__":
    sol = Solution()
    assert sol.minMeetingRooms([Interval(0, 40), Interval(5, 10), Interval(15, 20)]) == 2
    print("Test Case 1 passed.")
    assert sol.minMeetingRooms([Interval(4,9)]) == 1
    print("Test Case 2 passed.")
    print("\nAll Tests Passed.")
````
