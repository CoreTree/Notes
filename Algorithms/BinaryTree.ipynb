# Binary Trees
## Traversing Trees

```python
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

tree = TreeNode(val=1)
tree.left = TreeNode(val=2)
tree.left.left = TreeNode(val=4)
tree.left.right = TreeNode(val=5)
tree.right = TreeNode(val=3)
tree.right.left = TreeNode(val=6)
tree.right.right = TreeNode(val=7)
```

### Preorder Traversal of Binary Tree
Essentially, we will be navigating from top to bottom, starting from the left-most path before traveling to the right-path child nodes.

```python
def preorder(root:TreeNode) -> list[int]:
    res = []
    if root:
        res.append(root.val)
        res += preorder(root.left) + preorder(root.right)
    return res

print("Preorder:", preorder(tree))
```

### Postorder Traversal of Binary Tree
This involves going to the depths of the bottom-most left node and then going back up the right side involving all nodes before adding the root to the end of the traversal.

```python
def postorder(node:TreeNode) -> list[int]:
    res = []
    if node.left:
        res += postorder(node.left)
    if node.right:
        res += postorder(node.right)
    res.append(node.val)
    return res
print("Postorder:", postorder(tree)) #Referring to `tree` from the initial definition above
```

### Level Order Traversal of Binary Tree
Each path (left and right nodes) are all grouped as a level, and the main goal is to return each of these in this order. If you went 2 rows down, you'd be looking at the combination of `root.left.left`, `root.left.right`, `root.right.left`, and `root.right.right` nodes in your queue at that level.

```python
from collections import deque

def levelorder(node:TreeNode) -> list[int]:
    res = []
    q = deque()
    q.append(node)
    while len(q) > 0:
        cur = q.popleft()
        res.append(cur.val)
        if cur.left:
            q.append(cur.left)
        if cur.right:
            q.append(cur.right)
    return res

print("Levelorder:", levelorder(tree))
```

### Inorder Traversal of Binary Tree
Starting from the bottom-left-most node and working our way back up the stack and to the right-most path, we will return the path described.

```python
def inorder(node:TreeNode) -> list[int]:
    res = []
    if node.left:
        res += inorder(node.left)
    res.append(node.val)
    if node.right:
        res += inorder(node.right)
    return res

print("Inorder:", inorder(tree))
```
