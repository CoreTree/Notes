### Letter Combinatins of a Phone Number
Given a string of digits from 2-9 inclusive, return all possible letter combinations that the number could represent the answer in any order.

A mapping of digits to letters (just like on the telephone buttons) is given below. Note that 1 does not map to any letters.
![]()

#### Example 1
```
Input: digits = "23"

Output: ["ad", "ae", "af", "bd", "be", "bf", "cd", "ce", "cf"]
```
#### Example 2
```
Input: digits = ""

Output: []
```
#### Example 3
```
Input: digits = "2"

Output: ["a", "b", "c"]

````python
class Solution:
    def letter_combinations(self, digits:str) -> list[str]:
        if not hasattr(self, "lookup"):
            lookup = {
                "2": ["a", "b", "c"],
                "3": ["d", "e", "f"],
                "4": ["g", "h", "i"],
                "5": ["j", "k", "l"],
                "6": ["m", "n", "o"],
                "7": ["p", "q", "r", "s"],
                "8": ["t", "u", "v"],
                "9": ["w", "x", "y", "z"]
            }
        if len(digits) == 0:
            return []
        if len(digits) == 1:
            return lookup[digits[0]]
        res = []
        for letter in lookup[digits[0]]:
            for ltr in self.letter_combinations(digits[1:]):
                res.append(letter + ltr)
        return res
````

````python
def test_testcase(numb:int, digits:str, expected:list[str]):
    sol = Solution()
    res = sol.letter_combinations(digits)
    print(f"Test Case {numb}:")
    print("\tResults:", res)
    assert res == expected
    print("\tExpected:", expected)
````

````python
test_case_1_digits = "23"
test_case_1_expected = ["ad", "ae", "af", "bd", "be", "bf", "cd", "ce", "cf"]
test_testcase(1, test_case_1_digits, test_case_1_expected)

test_case_2_digits = ""
test_csae_2_expected = []
test_testcase(2, test_case_2_digits, test_csae_2_expected)

test_case_3_digits = "2"
test_case_3_expected = ["a", "b", "c"]
test_testcase(3, test_case_3_digits, test_case_3_expected)
````
