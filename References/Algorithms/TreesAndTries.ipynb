# Trees and Tries
This is a sample of how to navigate creating, traversing, and rebuilding a Tree and/or a Trie.

## Creating a Tree
### Using Preorder Tree Traversal

```python
# Start with creating a TreeNode
class TreeNode:
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

```python
def preorder_traversal(root:TreeNode) -> list[int]:
    res = []
    if root != None:
        res.append(root.val)
        res += preorder_traversal(root.left) + preorder_traversal(root.right)
    return res

node = TreeNode(val=1, left=TreeNode(val=2,
                            left=TreeNode(val=4),
                            right=TreeNode(val=5)),
                        right=TreeNode(val=3,
                            left=TreeNode(val=6),
                            right=TreeNode(val=7)))

if __name__ == "__main__":
    assert preorder_traversal(root=node) == [1, 2, 4, 5, 3, 6, 7]
    print("Preorder Traversal assertion passed!")
```

### Using Level Order Tree Traversal

```python
def level_order_traversal(root:TreeNode) -> list[int]:
    res = []
    q = [root]
    while len(q) != 0:
        temp = q.popleft()
        #In case root == None, we'd like to drop that one
        if temp == None:
            continue
        if temp.left:
            q.append(temp.left)
        if temp.right:
            q.append(temp.right)
        res.append(temp.val)
    return res

if __name__ == "__main__":
    assert level_order_traversal(root=node) == [1, 2, 3, 4, 5, 6]
    print("Level Order Traversal Passed!")
```

### Using Postorder Tree Traversal

```python
def postorder_traversal(root:TreeNode) -> list[int]:
    res = []
    if root == None:
        return res
    res += postorder_traversal(root.left)
    res += postorder_traversal(root.right)
    res.append(root.val)
    return res
```

### Using Inorder Tree Traversal

```python
def inorder_traversal(root:TreeNode) -> list[int]:
    res = []
    def recurse(node:TreeNode) -> list[int]:
        rec_res = []
        if node == None:
            return rec_res
        if node.left != None:
            rec_res += recurse(node.left)
        if node.right != None:
            rec_res += recurse(node.right)
        rec_res += [node.val]
        return rec_res
    if root == None:
        return res
    res += inorder_traversal(root.left)
    res += inorder_traversal(root.right)
    res += [root.val]
    return res
```

### Checking if it's the same tree (recursive)

```python
def compare(node_a:TreeNode, node_b:TreeNode) -> bool:
    if node_a and not node_b:
        return False
    if node_b and not node_a:
        return False
    if not node_a and not node_b:
        return True
    if node_a.val != node_b.val:
        return False
    return compare(node_a.left, node_b.left) and compare(node_a.right, node_b.left)
```

## Checking if it's the same tree (procedural)

```python
def compare_proc(node_a:TreeNode, node_b:TreeNode) -> bool:
    a_queue = [(node_a, "")]
    b_queue = [(node_b, "")]
    while a_queue and b_queue:
        if a_queue[0][1] != b_queue[0][1]:
            return False
        a_left, a_traversed = a_queue.popleft()
        b_left, b_traversed = b_queue.popleft()
        if a_left.val != b_left.val:
            return False
        if a_left.left:
            a_queue.append([a_left.left, a_traversed + "L"])
        if a_left.right:
            a_queue.append([a_left.right, a_traversed + "R"])
        if b_left.left:
            b_queue.append([b_left.left, b_traversed + "L"])
        if b_left.right:
            b_queue.append([b_left.right, b_traversed + "R"])
    return True
```
