# HashMaps

```python
# Valid Anagrams
def is_anagram(s:str, t:str) -> bool:
    s_dict = {}
    for ch in s:
        s_dict[ch] = s_dict.get(ch, 0) + 1
    t_dict = {}
    for ch in t:
        t_dict[ch] = t_dict.get(ch, 0) + 1
    for ch in t_dict:
        if ch not in s_dict:
            return False
        if s_dict[ch] != t_dict[ch]:
            return False
        del s_dict[ch]
    return len(s_dict) == 0
```
